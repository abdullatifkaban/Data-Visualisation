
**Gözlemleyebileceğin üç sonuç:**

3. Windows'ta Microsoft Store sayfası açıldı → Sistem Python'u Store'a yönlendiriyor; bu pencereyi kapat.

## 📖 Kavramlar

### Python dağıtımı nedir?

Python bir programlama dilidir; ama günlük kullanımda "Python kurmak" deyince Python yorumlayıcısı artı faydalı kütüphanelerden oluşan bir paket anlaşılır. Böyle paketlere **dağıtım (distribution)** denir. **Anaconda**, veri bilimi odaklı en yaygın dağıtımdır: içinde Python, Jupyter Notebook, pandas, matplotlib, seaborn gibi araçlar önceden kuruludur ve hepsi birbiriyle uyumlu sürümlerden seçilmiştir.

### Jupyter Notebook nedir?

**Jupyter Notebook**, kod, açıklama metni ve grafikleri aynı dosyada birleştiren bir ortamdır. Şu anda okuduğun dosya bir Jupyter defteri; açıklamalar markdown hücrelerinde, kodlar kod hücrelerinde duruyor ve kodlar çalıştırıldığında çıktıları hücrelerin altında beliriyor.

### Hücre ve çekirdek

- **Hücre (cell):** Defteri oluşturan her kutu. İki temel türü var: **kod hücresi** (Python kodu çalıştırır) ve **markdown hücresi** (başlık, metin, liste barındırır; bu hücre gibi).
- **Çekirdek (kernel):** Kodunu çalıştıran arka plandaki Python süreci. Defteri kapatmadan çekirdeği yeniden başlatabilirsin; bu, değişkenleri sıfırlar ama dosyayı bozmaz.

### Çalıştırma sırası

Kod hücrelerinin solundaki `In [1]`, `In [2]` ... numaraları çalıştırılma sırasını gösterir. Jupyter, hücreleri dosyadaki sırayla değil, **senin çalıştırdığın sırayla** işler. Bu esneklik güçlüdür ama dikkat istemezse hata çıkarır; "Deneyelim" bölümünde bunu canlı göreceğiz.

## 🧪 Deneyelim ve Gözlemleyelim

Şimdi Jupyter'in en çok kafa karıştıran özelliğini bilinçli olarak tetikleyelim: **çalıştırma sırası.** Kendi defterinde şu iki hücreyi oluştur:

```python
x = 5
```

```python
print(x)
```

**Deney 1:** Önce üstteki hücreyi, sonra alttakini çalıştır. `In` numaralarına ve çıktıya bak.

**Deney 2:** Şimdi üstteki hücreyi silip (hücre seçiliyken `d d` tuşlarına iki kez bas) yalnızca `print(x)` hücresini yeniden çalıştır. Ne olacak, tahmin et.

**Beklenen gözlem (Deney 2):** Şuna benzer bir hata alırsın:

```text
NameError: name 'x' is not defined
```

Neden? Çünkü `x` değişkeni çekirdeğin belleğinde, üstteki hücre çalıştırıldığı anda oluştu; hücreyi silmek belleği silmez ama çekirdeği **yeniden başlatırsan** (`Kernel → Restart Kernel`) tüm değişkenler gider. Bu deneyin dersi: defteri okuyan biri hücreleri dosya sırasıyla yukarıdan aşağıya çalıştırdığında her şey yolunda gider; sırayı bozarsan sonuçlara güvenemezsin.

**Deney 3:** Menüden `Kernel → Restart Kernel and Run All Cells` seçeneğini dene. Bu, "temiz sayfayla uçtan uca çalıştır" demektir ve defterini bitirdiğinde her zaman yapmanı önerdiğimiz kontrol adımıdır.

## 🧯 Bir Şeyler Ters Giderse

| Belirti | Olası neden | Çözüm |
|---|---|---|
| Terminalde `jupyter: command not found` | Jupyter yol (PATH) üzerinde değil | Windows'ta "Anaconda Prompt"u kullan; macOS/Linux'ta kurulumun ardından terminali kapatıp aç |
| Navigator açılıyor ama çok yavaş | Navigator arayüzü ağırdır | Terminalden `jupyter notebook` ile başlat (Yol B) |
| Tarayıcıda sayfa açılmıyor (`localhost:8888` boş) | Tarayıcı ya da port sorunu | Terminal çıktısındaki alternatif adresi (ör. `localhost:8889`) dene; tarayıcıyı değiştir |
| `import pandas` hata veriyor | Yanlış çekirdek/ortam seçilmiş | Defterin sağ üstündeki çekirdek adını kontrol et; "Python 3 (ipykernel)" olmalı |
| `Kernel not found` hatası | Deftere bağlı çekirdek yok | Çekirdek seçicisinden Python 3'ü yeniden seç |

> **İpucu:** Karşılaştığın hata mesajını sonuna kadar oku: son satırda hata türü (`NameError`, `ModuleNotFoundError` gibi) ve nedeni yazar. Bu dersteki tablo, ileride karşılaşacağın mesajların ilk haritası; 4. derste hata mesajı okumayı ayrıntılı öğreneceğiz.
