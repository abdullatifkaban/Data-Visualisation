## 📖 Kavramlar

**Değişken (variable):** Değere verdiğin addır. `not_ortalamasi = 84.6` satırı "not_ortalamasi adlı kutuya 84.6 koy" demektir. Eşittir işareti burada matematikteki "eşittir" değil, **atama**dir: sağdaki değer soldaki ada yazılır.

**Metin ve sayı:** `"Ayşe"` çift tırnak içinde → metin (Python'da `str`). `84.6` → noktalı sayı (`float`); `84` → tam sayı (`int`). Tür önemli: `"5"` bir metindir, `5` bir sayıdır; ikisi farklı davranır (aşağıda buna uygun bir kaza yaşayacağız).

**Liste (list):** Birden çok değeri tek adla sıralı tutar: `meyveler = ["elma", "armut", "kiraz"]`. Köşeli parantez, öğeler virgülle ayrılır. Öğeler sıralıdır ve numaraları sıfırdan başlar.

**Fonksiyon (function) ve argüman:** Fonksiyon, hazır bir işi yapan makinedir. `round(84.6)` → round fonksiyonuna 84.6 **argümanını** veriyoruz, bize 85 döndürüyor. Parantez içindeki her değer bir argümandır.

**Metot (method):** Nokta gösterimiyle çağrılan, belirli türe ait fonksiyonlar: `meyveler.append("muz")` listeye öğe ekler. `pd.read_csv(...)` gibi kodlarda göreceğin nokta da aynı mantıktır: "pd içindeki read_csv makinesi".

**Modül ve import:** Python'un tamamı tek paket değildir; ek yetenekler modüllerde gelir. `import math` deyince `math.sqrt(25)` ile karekök alabilirsin. Kurstaki `import pandas as pd` ve `import matplotlib.pyplot as plt` tam olarak budur: uzun adlara kısa takma ad veriyoruz.

### 2) Listeler: indeksleme ve uzunluk

````python
grafik_turleri = ["çizgi", "çubuk", "histogram", "kutu", "dağılım"]

print("Liste:", grafik_turleri)
print("İlk öğe:", grafik_turleri[0])
print("Son öğe:", grafik_turleri[-1])
print("Kaç öğe:", len(grafik_turleri))

grafik_turleri.append("ısı haritası")
print("Eklemeden sonra:", grafik_turleri)
````

`len` bir fonksiyon, `append` bir metot: ilki parantezle listeyi içine alıyor, ikincisi nokta ile kullanılıyor. İkisi de listeyle çalışıyor ama çağrı biçimleri farklı.

````python
siparisler = [120, 135, 150, 148, 170]

print("Toplam:", sum(siparisler))
print("En büyük:", max(siparisler))
print("En küçük:", min(siparisler))
print("Ortalama:", sum(siparisler) / len(siparisler))
print("Yuvarla:", round(104.6), round(104.4))
````

### 4) Metin ve sayı turu

````python
fiyat_metin = "250"
fiyat_sayi = 250

print(fiyat_metin * 2)    # metni ikiye katla?
print(fiyat_sayi * 2)     # sayıyı ikiye katla
print(fiyat_metin + " TL")
````

Aynı `*` ve `+` işleçleri türle anlam değiştiriyor: metinde çarpma "yan yana tekrarla", toplama "birleştir". Bu yüzden hangi türle çalıştığını bilmek önemli.

## 🧯 Bir Şeyler Ters Giderse

Bir hata çıktısına (traceback) şöyle bakılır: **en alttan yukarıya.** Son satır hata türünü ve açıklamayı taşır; üstündeki satırlar hatanın nerede olduğunu gösterir.

| Hata türü | Anlamı | İlk kontrolün |
|---|---|---|
| `NameError` | Tanımsız ad | Adın yazımını ve hücrenin çalıştırılmış olmasını |
| `TypeError` | Tür uyuşmazlığı | Tırnak işaretlerini (metin mi sayı mı) |
| `IndexError` | İndeks dışarıda | `len(...)` ile uzunluğu; son öğe `[-1]` |
| `ModuleNotFoundError` | Modül yok | `import` adını; Anaconda ortamında mısın |

> **İpucu:** Hata mesajını internette aramak da çözümün parçasıdır; hata türünü ve satırını aynen yapıştır. Çoğu kez aynı hatayı yaşamış birileri olmuştur.
