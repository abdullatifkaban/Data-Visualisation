## 🔍 Birlikte Keşfedelim

Kahve zincirinin satış kayıtları `../Data/05-coffee-sales.csv` dosyasında. Defterden bir üst klasöre çıkıp `Data` klasörüne giden bu yol, dosyanın deftere göre konumunu anlatır (3. dersteki not: veri dosyası ile defter aynı repoda ama farklı klasörlerde).

**Tahminin:** Dosyayı açmadan önce tahmin et: Bu dosyada hangi sütunlar olabilir? Kaç satır olabilir? (İpucu: zincir üç şehirde, üç ürün satıyor ve altı aylık kayıt tutuyor.) Tahminini bu hücreye yaz.

Şimdi yükleyip ilk beş satırına bakalım:

````python
import pandas as pd

satis = pd.read_csv("../Data/05-coffee-sales.csv")
satis.head()
````

**Gözlem:** Tabloda beş sütun var: `ay`, `sehir`, `urun`, `adet` ve `birim_fiyat`. Satır sayısını tahmin ettiysen kontrol edelim; etmediysen şimdi tahmin et:

````python
print("Satır sayısı:", satis.shape[0])
print("Sütun sayısı:", satis.shape[1])
print("Sütun adları:", list(satis.columns))
````

3 şehir × 3 ürün × 6 ay = 54 satır. **Şimdi asıl soru:** "Hangi şehir toplamda en çok satış yaptı?" Bu soruyu elle yanıtlamak 54 satırı gözden geçirmeyi gerektirir; pandas ile tek satır:

````python
satis.groupby("sehir")["adet"].sum().sort_values(ascending=False)
````

**Gözlem:** İstanbul açık ara önde, Ankara ve İzmir onu izliyor. Bu satırı şimdilik soldan sağa şöyle oku: `satis` tablosunu `sehir` sütununa göre grupla, her grubun `adet` toplamını al, büyükten küçüğe diz. Bunun nasıl çalıştığını şimdi parça parça açalım.

## 📖 Kavramlar

**DataFrame:** pandas'ın tablosu. Satırlar gözlemler, sütunlar değişkenler. Excel sayfasına benzer ama kodla sorgulanabilir.

**Sütun (Series):** DataFrame'in tek bir sütunu. `satis["adet"]` bize tüm satış adetlerini verir; listeden farkı, her değerin satır etiketiyle (indeks) taşınması.

**İndeks:** Satırların etiketi. Yüklediğimiz tabloda 0, 1, 2... biçiminde otomatik atandı; `head()` çıktısının solundaki koyu sayılar budur. Gruplama sonucunda indeks görevi değişir: yukarıdaki `groupby` sonucunda satır etiketleri şehir adları oldu.

**Veri türleri:** Sütunların türü `dtypes` ile görülür: metin sütunları `object`, tam sayılar `int64`, noktalı sayılar `float64`. Türler 6. derste (veri temizleme) başrolde olacak.

## 🛠️ Uygulama

### 1) Tabloyu tanıma

`head()` dışında üç hızlı tanıma komutu daha var. Her birini çalıştırmadan önce çıktısında ne göreceğini tahmin et:

````python
satis.info()
````

````python
satis.dtypes
````

````python
satis.describe()
````

`describe()` yalnızca sayısal sütunlar için özet üretir: sayı (count), ortalama (mean), standart sapma (std), en küçük/en büyük (min/max) ve çeyrekler. `birim_fiyat`'ın ortalamasının 86.67 gibi bir değer çıkması seni şaşırtmasın: üç ürünün fiyatı (90, 110, 60) her satırda tekrarlıyor.

### 2) Sütun seçme

````python
adetler = satis["adet"]          # tek sütun -> Series
print(adetler.head())
print("Ortalama:", adetler.mean())
````

````python
satis[["sehir", "urun", "adet"]].head()    # birden çok sütun -> yeni DataFrame
````

Tek köşeli parantez + tek ad = Series; çift köşeli parantez + ad listesi = yeni bir DataFrame. Fark küçük görünüyor ama karıştırmak yaygın bir hata kaynağı.

### 3) Koşulla filtreleme

````python
izmir = satis[satis["sehir"] == "İzmir"]
print("İzmir satır sayısı:", izmir.shape[0])
izmir.head()
````

````python
# iki koşul birden: Mayıs ayındaki Latte satışları
mayis_latte = satis[(satis["ay"] == "Mayıs") & (satis["urun"] == "Latte")]
mayis_latte
````

Koşul mantığı: `satis["sehir"] == "İzmir"` altı satır için True/False üretir; DataFrame'i köşeli parantezle süzünce yalnız True olan satırlar kalır. Birden çok koşulda her koşul paranteze alınır ve `&` (ve) ya da `|` (veya) ile birleştirilir — 4. dersteki `and`/`or` kelimeleri burada işlemez.

### 4) Sıralama

````python
# en yüksek tek satış kayıtları
satis.sort_values("adet", ascending=False).head()
````

### 5) Gruplama ve özetleme

Gruplama, "her ... için" sorusunun pandas karşılığıdır: her şehir için, her ürün için, her ay için...

````python
urun_toplam = satis.groupby("urun")["adet"].sum().sort_values(ascending=False)
urun_toplam
````

````python
# ortalamalar da alabiliriz: şehir bazında ortalama satış adedi
satis.groupby("sehir")["adet"].mean().round(1)
````

### 6) Yeni sütun türetme

Her satırın cirosunu (adet × birim fiyat) hesaplayıp tabloya yeni sütun olarak ekleyelim:

````python
satis["ciro"] = satis["adet"] * satis["birim_fiyat"]
satis.head()
````

````python
satis.groupby("urun")["ciro"].sum().sort_values(ascending=False)
````

Filtreleme + gruplamayı birleştirip zincir kurabiliriz. Örneğin yalnızca İstanbul'un ürün bazında cirosu:

````python
satis[satis["sehir"] == "İstanbul"].groupby("urun")["ciro"].sum()
````

## 🧪 Deneyelim ve Gözlemleyelim

**Deney 1:** Şehir adını küçük harfle yazarsan ne olur? Önce tahmin et, sonra çalıştır:

````python
satis[satis["sehir"] == "izmir"]
````

**Gözlem:** Hata yok, ama satır da yok: **boş bir tablo**. pandas, "izmir" diye bir şehir bulamadı ama bunu sana söylemedi. Bu tür **sessiz sonuçlar** hatalardan daha tehlikelidir; filtre sonucu beklenenden kısaysa koşulu (ve yazımı) gözden geçir. 6. derste bu tür sorunları sistematik yakalayacağız.

**Deney 2:** Parantezleri unutursan ne olur?

````python
satis["adet" > 100]
````

**Gözlem:** `TypeError`. Parantezleri karıştırınca Python önce `"adet" > 100` değerlendirdi; metin ile sayı arasında `>` kıyaslanamayınca hata verdi. Doğrusu koşulun tamamını paranteze almak: `satis[satis["adet"] > 100]`. Deneye devam edip doğrusunu çalıştır:

````python
satis[satis["adet"] > 500].head()
````

**Deney 3:** `ascending` parametresini kaldırırsan sıralama hangi yöne döner? Tahmin et, dene, `sort_values("adet")` yazarak kontrol et.

## 🧯 Bir Şeyler Ters Giderse

| Belirti | Olası neden | Çözüm |
|---|---|---|
| `FileNotFoundError` | Dosya yolu yanlış | Defterin konumuna göre yolu yaz: `../Data/...`; klasör ve dosya adının birebir aynı olduğundan emin ol |
| `KeyError: 'adt'` | Sütun adı yanlış yazılmış | `list(satis.columns)` ile gerçek adlara bak; büyük-küçük harf ve Türkçe karakterlere dikkat |
| Filtre boş tablo döndürdü | Koşul hiç satırla eşleşmedi | Değerleri `satis["sehir"].unique()` ile kontrol et |
| `TypeError` (koşul filtresinde) | Parantezler eksik ya da `and/or` kullanıldı | Koşulu paranteze al, `&` / `|` kullan |
| `NameError: 'pd'` | `import pandas as pd` çalıştırılmamış | İmport hücresini yeniden çalıştır |

> **İpucu:** pandas hataları genelde mesajın son satırında hem türü hem de asıl sorunu söyler. `KeyError`'da tırnak içinde yazdığı ad, senin yazdığın adla karşılaştırılınca çoğu zaman yazım hatası kendini ele verir.

## ✅ Kısa Özet

- DataFrame, kodla sorgulanan tablodur: satırlar gözlem, sütunlar değişkendir.
- Tanıma araçları: `head()`, `shape`, `columns`, `dtypes`, `info()`, `describe()`.
- Seçme: `satis["adet"]` (Series), `satis[["sehir", "adet"]]` (DataFrame).
- Filtreleme: `satis[satis["sehir"] == "İzmir"]`; koşullar parantezle ve `&`/`|` ile birleşir.
- Sıralama: `sort_values("adet", ascending=False)`.
- Gruplama: `groupby("...")["..."].sum()` — "her ... için" sorusunun karşılığı.
- Yeni sütun: mevcut sütunlardan hesaplayıp atanır: `satis["ciro"] = ...`.
- Boş tablo dönen sessiz sonuçlar, hata veren komutlardan daha sinsi olabilir.
