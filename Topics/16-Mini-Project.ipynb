- **Opsiyonel uzantı — streamlit panosu:** Analizini canlı sayfaya çevirmek istersen:

```python
# panom.py — bu kod defterde değil, ayrı bir .py dosyasında durur.
# Terminalden: pip install streamlit   sonra:   streamlit run panom.py
import streamlit as st
import pandas as pd

st.title("Öğrenci Alışkanlıkları Panosu")
veri = pd.read_csv("Data/10-student-habits.csv")
st.line_chart(veri.groupby("bolum")["vize_notu"].mean())
st.scatter_chart(veri, x="calisma_saati", y="vize_notu")
```
